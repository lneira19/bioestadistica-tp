```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import os
```

```python
repo_path = os.path.dirname(os.path.abspath(""))
dataset = pd.read_csv(os.path.join(repo_path, "dbs", "DatasetAfricaMalaria.csv"))
```

```python
dataset
```

```python
dataset.info()
```

```python
df_selected = dataset[[
    "Year", 
    "Country Name", 
    "Incidence of malaria (per 1,000 population at risk)", 
    "Urban population (% of total population)",
    "People using at least basic drinking water services (% of population)",
    "People using at least basic sanitation services (% of population)"]]
df_selected
```

```python
df_selected = df_selected.fillna("NaN")
```

```python
df_test = df_selected[(df_selected["Incidence of malaria (per 1,000 population at risk)"] == "NaN")]
countries = df_test["Country Name"].unique()
countries
```

```python
df_selected.info()
```

```python
df_selected_filtered = df_selected[~df_selected["Country Name"].isin(countries)].reset_index(drop=True)
df_selected_filtered.info()
```

```python
df_selected_filtered
```

```python
d_year = {f"{y}": df_selected_filtered[df_selected_filtered["Year"] == y] for y in df_selected_filtered["Year"].unique()}
for year, df_year in d_year.items():

    n = df_year["Country Name"].nunique()
    print(f"Año: {year} - Cantidad de países: {n}")

    # Averiguar si existe un 'NaN' en el dataframe
    if (df_year["Incidence of malaria (per 1,000 population at risk)"] == "NaN").any():
        print(df_year["Year"].iloc[0])
```
